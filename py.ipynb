```python
""" Question -2  largest number in lst """

def maximum_num(lst):
    max_num = lst[0]
    for i in lst:
        if i > max_num:
            max_num = i
    return max_num
lst_data = [1,2,3,4,9]



print(f" maximum number in list : {maximum_num(lst_data)}")
```

```python
""" Another Question - 16 : merged 2 list into dictionary """

def merged_lst(lst1,lst2):
    merged_lst = {i :j for i in lst1 for j in lst2 }
    return merged_lst

lst_1 = ['a','b','c']
lst_2 =[1,2,3]

print(merged_lst(lst_1,lst_2))
    
```

```python

""" Counting Frequency of words """

def frequency(words):
    count = {}
    words = words.split()
    for i in words:
        i = i.lower()
        count[i] = count.get(i,0)+1
    
    return count
sentence = " you are Very handsome You"
print(frequency(sentence))
```
