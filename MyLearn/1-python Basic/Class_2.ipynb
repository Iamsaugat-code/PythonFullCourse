```python
#variable checking 
first_name = "Ram"
last_name = "karki h"
age = 45
is_old = True
print(first_name,type(first_name))
print(last_name,type(last_name))
print(age,type(age))
print(is_old,type(is_old)) # and also checking the variable type in code
```

```python
# now checking for changing one type variable to another type of variable and 
# variable is sensitive 

age = 34
age_str = str(age)
print(age_str, type(age_str))
```

```python
#now checking for string variable and 
#we cannot convert full string into int e.g "hello" into int or float or double
name = '32'
name_int = int(name)
print(name_int,type(name_int))
```
