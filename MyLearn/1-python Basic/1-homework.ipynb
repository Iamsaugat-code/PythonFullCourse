**Question 17:** Write a Python program to check if a year is a leap year.

```python

# Your code here

year = int(input(" Enter the Year "))
if year %4 ==0:
    if year%100 ==0:
        if year%400:
            print(f"{year} is the leap Year ")
        else:
            print(f"{year} is not the leap Year ")
    else:
        print(f"{year} is the leap Year ")
else:
    print(f" {year} is no the leap year ")
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here


num = int(input(" Enter the Number "))
rev =0
orginal = num
while num >0:
    rem = num%10
    rev = rev*10+rem
    num = num//10
    
if rev == orginal:
    print(f"{orginal} is the Palindrom number ")
else:
    print(f"{orginal} is not the Plaindrom Number")
```
