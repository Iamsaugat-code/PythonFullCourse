### Magic method in Python

Magic methods are special built-in methods in Python that have double underscores (__) before and after their names. They allow objects to work with built-in functions, operators, and Python features.

```python
#### Example of magic method :
"""
__init__()	Initializes an object
__str__()	Returns a readable string for an object
__len__()	Returns the length of an object
__add__()	Defines what + does
__eq__()	Defines what == does
__repr__()	Returns the official representation of an obje
__getitem__()	Gets an item using an index or key
__setitem__()	value	Sets an item

"""
```

```python

## Common example of magic method

class Person:
    
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"{self.name} is a {self.age} years old : "
    
    def __repr__(self):
        return f"name : {self.name}, age : {self.age}"
    

p = Person("Ram Karki",32)
print(p)
print(repr(p))
```

```python

## length 

class People:
    def __init__(self,name):
        self.name = name
    
    def __len__(self):
        return len(self.name)

pe = People("asmitaRaidon")

print(len(pe))
print("list of magic method : ")
dir(pe)
```

#### Operator overloading 
Operator overloading is a feature in Python that allows operators to perform different operations based on the objects or data types being used.

```python
"""
__add__()	Adds two objects
__sub__()	Subtracts one object from another
__mul__()	Multiplies two objects
__truediv__()	Divides two objects
__eq__()    Checks whether two objects are equal
__gt__()    Checks whether the left object is greater
"""
```

```python

## Simple Example vector

class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x,self.y + other.y)
    
    def __sub__(self, other):
        return Vector(self.x - other.x , self.y - other.y)
    
    def __mul__(self, other):
        return Vector(self.x * other.x , self.y * other.y)
    
    def __eq__(self, other):
        return Vector(self.x == other.x ,self.y == other.y )
    
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    

v1 = Vector(2,3)
v2 = Vector(4,5)

print(v1+v2)
print(v1-v2)
print(v1*v2)
print(v1==v2)
```

```python

# complex Number 

class Complexed:
    
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Complexed(self.x + other.x,self.y+other.y)
    
    def __sub__(self, other):
        return Complexed(self.x - other.x,self.y-other.y)
    
    def __mul__(self, other):
        return Complexed(self.x*other.x-self.y*other.y,self.x*other.y + self.y*other.x)
    
    def __truediv__(self, other):
        deminator = other.x *other.x + other.y * other.y
        real = (self.x*other.x+self.y*other.y)/deminator
        img = (self.y*other.x - self.x*other.y)/deminator
        return Complexed(real,img)
    
    def __eq__(self, other):
        return Complexed(self.x == other.x,self.y == other.y )
    
    def __repr__(self):
        return f"Complex Number : {self.x} + {self.y}i"
    
    
c1 = Complexed(2,3)
c2 = Complexed(3,4)
print(c1+c2)
print(c1-c2)
print(c1*c2)
print(c1/c2)
print(c1==c2)

    
    
        
```

### Custom Execption handling

```python

class dobexecption(Exception):
    pass

year = int(input("Enter the year : "))
age = 2024-year

try:
    if age >20 and age <30:
        print("Be ready for the exam : ")
    else:
        raise dobexecption
except dobexecption:
    print("You are unable for the exam : ")
```

```python
## another example

class ilamExecption(Exception):
    pass

place = input("Enter the place : ")

try :
    if place == "ilam":
        print(" you are in right place : ")
    else:
        raise ilamExecption

except ilamExecption:
    print("You are in wrong place :")

except Exception as e:
    print(e)
```
