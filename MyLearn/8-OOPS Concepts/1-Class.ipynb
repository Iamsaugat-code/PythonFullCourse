### Class and Objects

Object Oriented Programming(OOP) Langauage is a programming paradigm that uses "object" to design application and computer program. OOPs allow for modeling real-world scenirors using class and objects.

```python
# Simple Example 

class car:
    pass

tata = car()
print(tata)

# note -- pass means do nothing inside the class 
```

```python

## instance variable and method 

class student:
    # __init__ means constructor in the python class
    
    def __init__(self,name,Class,grade):
        self.name = name
        self.Class = Class
        self.grade = grade
        
        # note -- self. means it represented the object of class
        
std1 = student("Supriya Limbu",3,"A+")
print("Memory Address : ",std1)
print("Name : ",std1.name)
print("Class : ",std1.Class)
print("Grade : ",std1.grade)
        


#  note -- while creating class (there is no any this )
        
```

```python
# Example of instance method

class Student:
    def __init__(self,name,age):
        self.hero = name
        self.life = age
    
    def about(self):
        return f"{self.hero} says to mamm : my age is {self.life}"

std1 = Student("Suman limbu",23)
print(std1.about())


# note -- while creating class use capital letter at first 
# note -- while creating any function we should give (self) in parameter
# note -- self make which function is calling 
```

```python
# Example Of bank account

class Bankaccount:
    
    def __init__(self,owner,balance = 0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self,amount):
        self.balance += amount
        
        print(f"{self.owner} has deposit Rs {amount} : New Balance is {self.balance}")
    
    def withdraw(self,amount):
        if amount > self.balance:
            print("Your amount is less : ")
        else:
            self.balance -= amount
            print(f"{self.owner} had withdraw Rs {amount} : New Balance : {self.balance}")
    
    def check_balance(self):
        print(self.balance)
        

p1 = Bankaccount("Ram Karki",5000)

print("Checking balance : ")
p1.check_balance()

print("\nDeposite : ")
p1.deposit(1000)

print("\nWithDraw : ")
p1.withdraw(5000)

print("\nlast Checked : ")
p1.check_balance()

    
```
