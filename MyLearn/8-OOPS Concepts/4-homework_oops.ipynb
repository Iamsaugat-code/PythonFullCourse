#### Assignment 1: Polymorphism with Methods

Create a base class named `Shape` with a method `area`. Create two derived classes `Circle` and `Square` that override the `area` method. Create a list of `Shape` objects and call the `area` method on each object to demonstrate polymorphism.

```python
import math

class Shape:
    
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius = radius
    
    def area(self):
        return f"area of Circle is {math.pi * self.radius*self.radius}"

class Square(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius = radius
    
    def area(self):
        return f"area of Square is {self.radius*self.radius}"

obj = [Circle(4),Square(4)]

for show in obj:
    print(show.area())
```

Create a function named `describe_shape` that takes a `Shape` object as an argument and calls its `area` method. Create objects of `Circle` and `Square` classes and pass them to the `describe_shape` function.

#### Assignment 3: Abstract Base Class with Abstract Methods

Create an abstract base class named `Vehicle` with an abstract method `start_engine`. Create derived classes `Car` and `Bike` that implement the `start_engine` method. Create objects of the derived classes and call the `start_engine` method.

```python

from abc import ABC, abstractmethod

class Vehicle:
    
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    
    def start_engine(self):
        print("Car Engine started to run : ")

class Bike(Vehicle):
    
    def start_engine(self):
        print("Bike engine started to run : ")

obj = [Car(),Bike()]

for show in obj:
    show.start_engine()
```

```python
from abc import ABC, abstractmethod

class Vehicle:
    
    @abstractmethod
    def start_engine(self):
        pass
    
    def fuel_type(self):
        return "Generic fuel : "

class Car(Vehicle):
    
    def start_engine(self):
        print("Car Engine started to run : ")
        
    def fuel_type(self):
        return "Generic Fuel from car : "

class Bike(Vehicle):
    
    def start_engine(self):
        print("Bike engine started to run : ")
        
    def fuel_type(self):
        return "generic fuel from bike : "

obj = [Car(),Bike()]

for show in obj:
    print(show.fuel_type())
```

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Ensure that the balance cannot be accessed directly.

```python

class BankAccount:
    
    def __init__(self,account_number,balance = 0):
        
        self.__account_no = account_number
        self.__balance = balance
        
        # private attribute 
    
    def deposit(self,amt):
        self.balance += amt
        print(f"{amt} has deposit : New Balance : {self.balance}")
    
    def withdraw(self,amt):
        self.balance -= amt
        print(f"{amt} has withdraw : New Balance : {self.balance}")
    
    def check(self):
        print(f"Balance : {self.balance}")

account = BankAccount(12344,5000)
account.deposit(3000)
account.withdraw(2000)
account.check()
        
```

#### Assignment 6: Encapsulation with Property Decorators

In the `BankAccount` class, use property decorators to get and set the `balance` attribute. Ensure that the balance cannot be set to a negative value.

```python
""" @property → used to get/read a value
 @name.setter → used to set/change a value
 @name.deleter → used to delete a value """
 


class negative_value(Exception):
    pass

class BankAccount:
    
    def __init__(self,ac_no,balance =0):
        
        self.__ac_no = ac_no
        self.__balance = balance
    
    @property
    def balance(self):
        return f"Your balance {self.__balance}" 
    
    @balance.setter
    def balance(self,amt):
        try:
            if amt >0:
                self.__balance += amt
            else:
                raise negative_value
        
        except negative_value:
            print("Negative value is not allowed : ")
    
    def deposit(self,amt):
        try:
            if amt>0:
                self.__balance += amt
            else:
                raise negative_value
        except negative_value:
            print("negative value is not deposit : ")
    
    def withdraw(self,amt):
        try:
            if amt>0:
                self.__balance -= amt
            else:
                raise negative_value
        except negative_value:
            print("negative value are not allowed : ")


u1 = BankAccount(122,450)
print(u1.balance)
u1.deposit(-3)
u1.withdraw(-3)
    
    
 
 
```

#### Assignment 9: Abstract Methods in Base Class

Create an abstract base class named `Employee` with an abstract method `calculate_salary`. Create two derived classes `FullTimeEmployee` and `PartTimeEmployee` that implement the `calculate_salary` method. Create objects of the derived classes and call the `calculate_salary` method.

```python

from abc import ABC,abstractmethod

class Employee:
    
    @abstractmethod
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):
    
    def __init__(self,salary):
        self.salary = salary
    
    def calculate_salary(self):
        return f"Full Time Employee Salary : {self.salary}"
    

class PartTimeEmployee(Employee):
    
    def __init__(self,salary):
        self.salary = salary
    
    
    def calculate_salary(self):
        return f"Part time employee salary : {self.salary}"

obj = [FullTimeEmployee(10000),PartTimeEmployee(5000)]

for show in obj:
    print(show.calculate_salary())
        
```

### Assignment 11: Polymorphism with Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python


class Vector:
    
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x,self.y + other.y)
    
    def __str__(self):
        return f"Vector({self.x},{self.y})"

v1 = Vector(2,3)
v2 = Vector(3,4)

print(v1+v2)
```

#### Assignment 12: Abstract Properties

Create an abstract base class named `Appliance` with an abstract property `power`. Create two derived classes `WashingMachine` and `Refrigerator` that implement the `power` property. Create objects of the derived classes and access the `power` property.

```python
"""power is a property
@abstractmethod makes it mandatory
Every child class must create its own power property"""


from abc import ABC, abstractmethod

class Appliance:
    
    @property
    @abstractmethod
    def power(self):
        print("Means : 1000 Hz")

class WashingMachine(Appliance):
    @property
    def power(self):
        return "power less than refrigeration : "

class Refrierator(Appliance):
    @property
    def power(self):
        return "power is greater than washingMachine : "

obj = [WashingMachine(),Refrierator()]

for show in obj:
    print(show.power)
    
    
```

#### Assignment 15: Abstract Methods and Multiple Inheritance

Create an abstract base class named `Worker` with an abstract method `work`. Create two derived classes `Engineer` and `Doctor` that implement the `work` method. Create another derived class `Scientist` that inherits from both `Engineer` and `Doctor`. Create an object of the `Scientist` class and call the `work` method.

```python

from abc import ABC,abstractmethod

class Worker:
    @abstractmethod
    def work(self):
        pass


class Engineer(Worker):
    def work(self):
        print("He is a worker : ")

class Doctor(Worker):
    def work(self):
        print("He is a doctor : ")

class  Scientist(Engineer,Doctor):
    pass

man = Scientist()
man.work()

# note --- Scientist(Engineer,Doctor) produce result on the basic first come
```
