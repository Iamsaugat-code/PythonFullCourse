### Inheritane In OOPS
Inheritance is a feature in Python where one class can use the properties and methods of another class.

```python

## Single Inheritance Example 

class Car :
    def __init__(self,window,doors,enginetype):
        self.window = window
        self.doors = doors
        self.enginetype = enginetype
    
    def Driver(self):
        print(f"A person is driving {self.enginetype} Car ")
        
        # note -- every function(there is self because it represent which function is calling : )

car1 = Car(4,4,"Petrol")
car1.Driver()

#  making Child Class

print("\nShowing From the Child Classn : \n")
class Tesla(Car):
    def __init__(self, window, doors, enginetype,is_selfdriving):
        super().__init__(self,window, doors, enginetype)
        self.is_selfdriving = is_selfdriving
    
    def selfDriving(self):
        print(f"Tesla is a self Driving Car :  {self.is_selfdriving}")

tesla1 = Tesla(4,4,"Electric",True)
tesla1.selfDriving()
tesla1.Driver()
```

```python

## Creating Multiple Inheritance 

# Base Class - 1
class Animal:
    def __init__(self,name):
        self.name = name
    
    def speaking():
        print("Dog is braking to the man : ")

# Base Class - 2 

class Pet:
    def __init__(self,owner):
        self.owner = owner

# Making child Class 

class Dog(Animal,Pet):
    
    def __init__(self, name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speaking(self):
        print(f"{self.name} is barking to the man")

#  making object

dog = Dog("Puppy","Ram Rai")
print()
dog.speaking()
print("Owner Name : ",dog.owner)
        
    

```

## Polymorphism
Polymorphism means “one thing, many forms.
In Python, the same method or function can behave differently depending on the object.

### Method Overriding
Method overriding means a child class provides its own version of a method that already exists in the parent class.

```python

# Example of Method Overriding

class Animal:
    def sound(self):
        return "Animal Sound ! "

# derived Class

class Cat(Animal):
    def sound(self):
        return "Meow ! "

# derived Class

class Dog(Animal):
    def sound(self):
        return "Bark ! "

#  make the demostrate function 

def animal(speak):
    print(speak.sound())

cat = Cat()
print(cat.sound())
dog = Dog()
print(dog.sound())
animal(cat)
```

```python

# Polymorphism with function and method 

class Shape:
    def area(self):
        return "This is the area of the Shape"

# Derived Class - 1 

class Reactangle(Shape):
    
    def __init__(self,lenght,width):
        self.length = lenght
        self.width = width
    
    def area(self):
        return self.length * self.width

# Derived Class - 2

class Circle(Shape):
    
    def __init__(self,side):
        self.side = side
    
    def area(self):
        return 3.14 * self.side **2

# demostrate the function 

def print_area(shape):
    print(f"Area of of shape : {shape.area()}")


reactangle = Reactangle(3,4)
circle = Circle(4)

print_area(reactangle)
print_area(circle)
        
    
```

## Ploymorphism with Abstract Base Class

plymorphism with an Abstract Base Class (ABC) means different child classes implement the same abstract method in their own way

##### Role of @abstractmethod 
It is used to declare a method that must be implemented by child classes.
here, sound() is an abstract method. Any child class of Animal must define its own sound() method:
If a child class does not implement sound(), Python will not allow you to create its object.

```python

from abc import ABC, abstractmethod

class Vechile:
    
    @abstractmethod
    def start_engine(self):
        pass
    
    # note --- pass means "do nothing for now "

class Car(Vechile):
    
    def start_engine(self):
        return "Car engine is started"

class Motorcycle(Vechile):
    
    def start_engine(self):
        return "MotorCycle engine is started"


def start_engine(vechile):
    print(vechile.start_engine())

car = Car()
motor = Motorcycle()

start_engine(car)
start_engine(motor)    
```

### Encapsulation and Abstraction 
Encapsulation in Python means wrapping data (variables) and methods (functions) together inside a class and controlling access to the data. Abstraction means hiding unnecessary implementation details and showing only the important features to the user. In short, encapsulation protects data, while abstraction hides complexity.

### Encapsulation
Encapsulation means combining data (variables) and methods (functions) into a single unit called a class and controlling how the data is accessed or modified.

```python

# Public Access 

class Person:
    
    def __init__(self,name,age):
        self.name = name
        self.age = age
    

p = Person("Ramesh Rai",34)
print(p.name)
print(p.age)

#  note -- in pulic access dir(p) show list of function with data
```

```python

## Private Access 

class Person:
    
    def __init__(self,name,age):
        
        self.__name = name    # private data
        self.__age = age      # private data 
    
    def showdata(self):
        print(self.__name)
        print(self.__age)

p = Person("Ram Limbu ",55)
p.showdata()


# note --- (__) if u give two this it's means it is private 

#  note - dir(p) does not show the list of function of data 5
#  '_Person__age',
#  '_Person__name',  but this is bad practice
        
```

```python

## protected Data 

class Person:
    
    def __init__(self,name,age):
        self._name = name   # protected data
        self._age = age     # protected data

class Student(Person):
    
    def __init__(self, name, age):
        super().__init__(name, age)


std = Student("Ravi Teja",44)

print("Name : ",std._name)
print("Age : ",std._age)
```

```python
## Getter and setter in private data

class Student:
    
    def __init__(self,name,age):
        self.__name = name
        self.__age = age
    
    # getter name
    
    def get_name(self):
        return self.__name
    
    # setter name
    
    def set_name(self,name):
        self.__name = name
    
    # getter age
    
    def get_age(self):
        return self.__age
    
    #  setter age
    
    def set_age(self,age):
        if age >=0:
            self.__age = age
        
        else:
            print("age cannot be less than zero")

s = Student("sabin rai",33)
print(s.get_name())
print(s.get_age())

s.set_age(45)
print(s.get_age())

s.set_age(-5)
    
```

### Abstraction 
Abstraction means hiding unnecessary internal details and showing only the important features to the user.

Example:
When you drive a car, you use the steering wheel, brake, and accelerator, but you do not need to know how the engine works internall

```python

## Using abstraction

from abc import ABC , abstractmethod

class Vehicle(ABC):
    
    def drive(self):
        print("Vehicle are used for dirving")
        
    @abstractmethod
    
    def start_engine(self):
        pass

class Car(Vehicle):
    
    def start_engine(self):
        print("Car Engine started to run : ")


def show_me(vehicle):
    vehicle.start_engine()

car = Car()
show_me(car)

```
