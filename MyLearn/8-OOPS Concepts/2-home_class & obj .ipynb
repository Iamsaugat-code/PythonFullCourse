### Assignment 11: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python

# Creating Abstract Class 

from abc import ABC , abstractmethod

class Shape:
    @abstractmethod
    # this means  -- every class can have own area() and can be used 
    def area():
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius**2

class Square(Shape):
    def __init__(self,side):
        self.side = side
    
    def area(self):
        return self.side ** 2

cir = Circle(6)
sqr = Square(4)

print(cir.area())
print(sqr.area())
        
```

### Assignment 12: Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python

class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"Vector X : {self.x}, Vector Y : {self.y}"

v1 = Vector(3,5)
v2 = Vector(3,1)
v3 = v1+v2
print(v3)

#  __str__ -- means it controls what is shown when you print an object using print().
```

### Assignment 13: Class with Custom Exception

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.

```python

class InsufficientBalanceError(Exception):
    pass
# note -- above class means creating own exception handling



class BankAccount:
    def __init__(self,balance  =0 ):
        self.money = balance
    
    def withdraw(self,amount):
        if amount > self.money:
            raise InsufficientBalanceError("! Insufficient Money : ")
        else:
            self.money -= amount
    
    def showbalance(self):
        print(f"Balance : {self.money}")

p1 = BankAccount(400)
print("Before WithDraw : ")
p1.showbalance()

try :
    p1.withdraw(200)
except InsufficientBalanceError as E:
    print(E)
else:
    print("\nAfter Withdraw")
    p1.showbalance()


  
#  Note -- used raise when you are using if condition and inside it exception 
        
```


### Assignment 14: Class with Context Manager

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python

class fileManager:
    def __init__(self,filepath,mode):
        self.filepath = filepath
        self.mode = mode
        
    #  making file open 
    def __enter__(self):
        self.file = open(self.filepath,self.mode)
        return self.file
    
    #  closing a file
    
    def __exit__(self, exc_type, exc, tb):
        return self.file.close()


with fileManager("example_1.txt",'r') as file:
    contents = file.read()
    print(contents)
```

### Assignment 15: Chaining Methods

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python

class Calculator:
    def __init__(self,value =0):
        self.value = value
    
    def add(self,amt):
        self.value += amt
        return self
    
    def sub(self,amt):
        self.value -= amt
        return self
    
    def mult(self,amt):
        self.value *= amt
        return self
    
    def divide(self,amt):
        if amt != 0:
            self.value /= amt
        else:
            print("Cannot divide by zero : ")
        return self

cals = Calculator()
cals.add(10).sub(2).mult(4).divide(2)
print(cals.value)

```
