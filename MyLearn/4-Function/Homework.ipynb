#### Assignment 1: Fibonacci Sequence with Memoization
Define a recursive function to calculate the nth Fibonacci number using memoization. Test the function with different inputs.

```python


""" using memoization for fibonacci series """

def fibo(n,memo={}):
    if n in memo:
        return memo[n]
    if n <=1:
        return n
    memo[n] = fibo(n-1,memo) + fibo(n-2,memo)
    return memo[n]

print("ans : ",fibo(8))
```

#### Assignment 6: Function with Decorators

Define a function that calculates the time taken to execute another function. Apply this decorator to a function that performs a complex calculation. Test the decorated function with different inputs.


```python

""" calculation time take for execute a program : """

import time

def time_decoration(func):
    def wrapper(*argu,**kwargu):
        start_time= time.time()
        result = func(*argu,**kwargu)
        end_start = time.time()
        print(f" Function {func.__name__} take {start_time-end_start} for excute : ")
        return result
    return wrapper

@time_decoration

def time_calculation(n):
    return sum(i**2 for i in range(n))

print(time_calculation(20000))


#  note 
# func(*k,**k) means = it will aspect (100 or a="1000")
# import time help to show time taken by the program 
# start = time.start : it show time started time
# end = time.time() : it stop the time for execuationg 

# func.__name__  = help to show the function name 

        
            
```

#### Assignment 8: Function Composition

Define a function that composes two functions, f and g, such that the result is f(g(x)). Test with different functions f and g.

```python

"""  making compose function  """

def compose(f,g):
    return lambda x : f(g(x))

f = lambda x:x+2
g = lambda x : x*2

show = compose(f,g)
print(show(2))
print(show(3))
#  note - after completing g function work than it enter into f function

```

#### Assignment 10: Function with Error Handling

Define a function that takes a list of integers and returns their average. The function should handle any errors that occur (e.g., empty list) and return None in such cases. Test with different inputs.


```python

def avgs(lst):
    try:
        return sum(lst)/len(lst)
    except ZeroDivisionError:
        return "No data is there"

print(avgs([1,2,3,4,5]))
print(avgs([]))
```

#### Assignment 13: Function with Context Manager

Define a function that uses a context manager to write a list of integers to a file. The function should handle any errors that occur during file operations. Test with different lists.

```python

#  making list data to write in file...

def file_to_write(filename,lst):
    try:
        with open(filename,'w') as f:
            for dat  in lst:
                f.write(f"{dat}\n")
    except IOError as E:
        print(E)


print(file_to_write("she.txt",["aaple","mango","hero"]))

```

```python
def separate_data(lst):
    Int =[]
    Float = []
    String = []
    for i in lst:
        if isinstance(i,int):
            Int.append(i)
        elif isinstance(i,float):
            Float.append(i)
        else:
            String.append(i)
    
    return Int,Float,String

mixed = [1,3.5,"shiva","rohan",6.7,2,3]

print(separate_data(mixed))
        
            
```
