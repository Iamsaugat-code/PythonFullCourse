
### Logging in Python 

Logging in Python is a way of recording information about what happens while a program runs. It helps us track errors, warnings, important events, and program activities, which makes debugging easier. Python provides a built-in logging module for this purpose. For example, we can use logging.info() for normal information, logging.warning() for warnings, and logging.error() for errors.

```python

# creating simple logging configuration 


import logging

logging.basicConfig(level =logging.DEBUG)

# create mss for to display 

logging.debug("This is the message for debuging : ")
logging.info("this is message for information :")
logging.warning("This is the message for warning :")
logging.error(" this is the message for error : ")
logging.critical(" this is the message for critical : ")

```


### level of logging 

##### DEBUG – Detailed information used mainly for debugging.
##### INFO – General information about the normal operation of the program.
##### WARNING – Something unexpected happened, but the program can continue.
##### ERROR – A serious problem occurred, and some part of the program may not work.
##### CRITICAL – A very serious error that may stop the entire program.

```python

#  creating message with format 

import logging

logging.basicConfig(level=logging.DEBUG,
                    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
                    datefmt='%Y-%m-%d - %H:%M:%S',
                    force=True
                    )


logging.debug("This is the message for debuging : ")
logging.info("this is message for information :")
logging.warning("This is the message for warning :")
logging.error(" this is the message for error : ")
logging.critical(" this is the message for critical : ")


# %m   # month
# %M   # minute
# %S   # second
# force=True means “force Python to apply my new logging settings,
# even if logging was already configured.”
```

```python

# creating another logging with fila name

import logging

logging.basicConfig(
   filename='app.log',
   filemode='w',
   level=logging.DEBUG,
   format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
   datefmt='%Y-%m-%d - %H:%M:%S'
)

logging.debug("This is the message for debuging : ")
logging.info("this is message for information :")
logging.warning("This is the message for warning :")
logging.error(" this is the message for error : ")
logging.critical(" this is the message for critical : ")


# we should do resart because old logging setting will not work
```


### Logging with multiple logger

```python

# multiple logger 

import logging

# creating logger - 1

logger1 = logging.getLogger("Module_1")
logger1.setLevel(level=logging.DEBUG)


# creating logger - 2

logger2 = logging.getLogger("Module_2")
logger2.setLevel(level=logging.WARNING)

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt =  '%Y-%m-%d  %H:%M:%S',
)

logger1.debug("This is the message of debug : ")
logger2.error("This is the message for error : ")
logger2.warning("This is the message for warning ")

```
