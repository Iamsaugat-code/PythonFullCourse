```python

from logging.handlers import RotatingFileHandler


def rotate_file_handler():
    
    logger = logging.getLogger("rotating file : ")
    logger.setLevel(level=logging.DEBUG)
    
    
    file_rotating = RotatingFileHandler("File.log",maxBytes=500,backupCount=6)
   
    
    format1 = logging.Formatter('%(asctime)s-%(name)s-%(levelname)s-%(message)s')
    file_rotating.setFormatter(format1)
    
    logger.addHandler(file_rotating)
    
    for i in range(100):
        logger.debug('This is the message for debug : {} '.format(i))
        
    
    
rotate_file_handler()
```

#### Assignment 6: Contextual Logging

1. Write a Python function to create a logger that includes contextual information (e.g., function name, line number) in the log messages.

```python


def context_log():
    
    logger = logging.getLogger("contextual")
    logger.setLevel(level=logging.DEBUG)
    
    file_handler = logging.FileHandler("custom.log")
    formating = logging.Formatter('%(asctime)s-%(name)s-%(levelname)s-%(message)s-%(funcName)s-%(lineno)d')
    file_handler.setFormatter(formating)
    
    logger.addHandler(file_handler)
    
    def myhome():
        
        logger.debug("This is the message for debug : ")
        logger.warning("This is the message for warnign : ")
        logger.error("This is the message for error : ")
        logger.info("This is the message for info : ")
        logger.critical("This is the message for critical : ")
    
    myhome()


context_log()


# "The value of lineno should be formatted as an integer (decimal number)." )d,)s
    
    
```

#### Assignment 7: Configuring Logging with a Dictionary

1. Write a Python function to configure logging using a dictionary. The configuration should include handlers for both file and console logging.

```python

import logging

import logging.config

def dict_logging():
    
    dict_log ={
        'version':1,
        'formatters':{
            'default':{
                'format':'%(asctime)s-%(name)s-%(levelname)s-%(message)s'
            },
            'detailed':{
                'format':'%(asctime)s-%(name)s-%(levelname)s-%(message)s-%(funcName)s-%(lineno)d'
            }
        },
        'handlers':{
            'file':{
                'class':'logging.FileHandler',
                'filename':'dict_log.log',
                'formatter':'detailed',
                'level':'DEBUG',
            },
            'console':{
                'class':'logging.StreamHandler',
                'formatter':'default',
                'level':'DEBUG'
            }
        },
        'root':{
            'handlers':['file','console'],
            'level':'DEBUG'
        }
    } 
    
    logging.config.dictConfig(dict_log)
    
    logger = logging.getLogger('')
    
    logger.debug("This is the message for debug : ")
    logger.warning("This is the message for warnign : ")
    logger.error("This is the message for error : ")
    logger.info("This is the message for info : ")
    logger.critical("This is the message for critical : ")

dict_logging()


# '' = This matches the 'root' configuration you created above.
```

#### Assignment 9: Logging Performance

1. Write a Python script to benchmark the performance of logging with different handlers (e.g., file handler, console handler, rotating file handler).

```python

import logging
import time
from logging.handlers import RotatingFileHandler


def benchmark_performance():
    
    logger = logging.getLogger("Bench_performance")
    logger.setLevel(level=logging.DEBUG)
    
    
    # file 
    filehander = logging.FileHandler("file.log")
    filehander.setLevel(level=logging.DEBUG)
    logger.addHandler(filehander)
    
    start = time.time()
    for _ in range(1000):
        logger.debug("this time calculating message : ")
    end = time.time()
    print('file handling time executin is : {}'.format(end-start))
    logger.removeHandler(filehander)
    
    
    # console
    consolehandler = logging.StreamHandler()
    consolehandler.setLevel(logging.DEBUG)
    logger.addHandler(consolehandler)
    
    start = time.time()
    for _ in range(1000):
        logger.debug("this console message for time calculation : ")
    end = time.time()
    print('console get time for execution is {}'.format(end-start))
    logger.removeHandler(consolehandler)
    
    
    # rotating file
    
    rotatingfile = RotatingFileHandler(filename='rotating.log',maxBytes=2000,backupCount=5)
    rotatingfile.setLevel(logging.DEBUG)
    logger.addHandler(rotatingfile)
    
    start = time.time()
    for _ in range(1000):
        logger.debug("rotating message for time calculation : ")
    end = time.time()
    print('rotating file will take : {} time '.format(end-start))
    logger.removeHandler(rotatingfile)
         

benchmark_performance()

```
