### Assignment 6: Working with the `os` Module

Use the `os` module to create a new directory, list the contents of the current directory, and remove the newly created directory.

```python
import os

# making new direction
print(os.makedirs("new_dir"))

#  show list of content present 
print(os.listdir('.'))

# note . listdir('.') will show the list list file present this folder

# removing new direction 
print(os.removedirs("new_dir"))
print(os.listdir('.'))
```

### Assignment 9: Working with the `datetime` Module

Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.

```python

import datetime

#  today date 

today = datetime.date.today()
print(f"Today : {today}")

#  Date after 100 Days

future = today+datetime.timedelta(days=100)
print(f"Date After 100 Days : {future}")

 
#  Day of week of the given data

given_day = datetime.date(2023,1,3)
print(f"Day of week of given Date : {given_day.strftime('%A')}")

#  note -- strftime('%A') it retrun full name of day
```

### Assignment 10: Working with the `random` Module

Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.

```python
import random

#  making 5 random number

random_5 = [random.randint(1,50) for _ in range(5)]

print(f"Five Random Number : {random_5}")

#  creating shuffle
lst = [1,2,3,4,5]
random.shuffle(lst)
print(lst)
```
