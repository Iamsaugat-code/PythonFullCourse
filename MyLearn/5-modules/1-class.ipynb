```python
''' converting dict to json and json to dict '''

import json

mydata=[{'name':'sabin rai','age':26},
        {'name':'sital rai','age':12},]

new_json = json.dumps(mydata)
print(new_json)
print(type(new_json))

#  again converting json into dict 

new_dict = json.loads(new_json)
print(new_dict)
print(type(new_dict))
```

```python
''' CSV Moduls '''

#  opening CSV file
 
import csv

with open('Example.csv',mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name','Age'])
    writer.writerow(['Krishna',32])

with open('Example.csv',mode='r') as file:
    read = csv.reader(file)
    for line in read:
        print(line)
    
        
```

```python
''' datatime moduls '''

from datetime import datetime,timedelta

now = datetime.now()
print(now)

yesterday = now - timedelta(days=1)
print(yesterday)
```
