### Understanding Execption 

Exception handling is the process of detecting and handling errors during program execution without stopping the entire program.
.Nameerror 
.Zerodivision error 
.valueError . invalid error 


```python
# NameErro

try:
    a=b
except NameError as ex:
    print(ex)
```

```python
# Division by zero 

try:
    a = 5/0
except ZeroDivisionError as e:
    print(e)
    print("Please used the number greater than zero : ")
```

```python
# Exception handling 

try:
    a = 1/2
    a =b
except ZeroDivisionError as e:
    print(e)
    print("Use the Number greater than zero :  ")
except Exception as ex:
    print(ex)
    print("Possible Error got catch : ")
```

```python
# Value Error

try :
    num = int(input("Enter the Number : "))
    result = 10/num
except ValueError as v:
    print(v)
    print("Enter the number only : ")
except ZeroDivisionError as z:
    print(z)
except Exception as ex:
    print(ex)
```

```python
# try,except,else

try :
    num = int(input("Enter Number : "))
    result = 10/num
except ValueError as v :
    print(v)
except Exception as ex:
    print(ex)
else:
    print(f"Ans : {result}")
     
```

```python
# try , except , else , finally 

try :
    num = int(input("Enter Number : "))
    result = 10/num
except ValueError as v :
    print(v)
except Exception as ex:
    print(ex)
else:
    print(f"Ans : {result}")
finally:
    print("Execution Completed : ")
```
