### Data visualization with Seaborn 
it is the process of representing data using different types of graphs and charts in Python. Seaborn is a Python library built on top of Matplotlib that makes it easier to create attractive and informative visualizations. It is commonly used for bar plots, histograms, scatter plots, box plots, heatmaps, and distribution plots. Seaborn works especially well with Pandas DataFrames and helps us understand patterns, relationships, and distributions in data.

```python

import seaborn as sbn
```

```python

# basic plotting with seaborn

tips = sbn.load_dataset('tips')
tips.head(5)
```

```python

#  making scatter plotting with seaborn data

import matplotlib.pyplot as plt

plt.figure(figsize=(5,4))
sbn.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot total Bill vs Tips")
plt.show()
```

```python

# line plotting

plt.figure(figsize=(5,4))
sbn.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line plotting of total_bill and days")
plt.show()
```

```python

#  bar plotting

plt.figure(figsize=(5,4))
sbn.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar plotting of day total_bill : ")
plt.show()
```

```python

# Box plotting

plt.figure(figsize=(5,4))
sbn.boxplot(x='day',y='total_bill',data=tips)
plt.title("Box plot of Day & Total_Bill")
plt.show()
```

```python

# violien plot 

plt.figure(figsize=(5,4))
sbn.violinplot(x='day',y='total_bill',data=tips)
plt.title("Violien plot of Day & total_bill")
plt.show()
```

```python

#  histogram plot

plt.figure(figsize=(5,4))
sbn.histplot(tips['total_bill'],bins=10,kde=True)
plt.title("Histogram total bill")
```

```python

# kde plot

plt.figure(figsize=(5,4))
sbn.kdeplot(tips['total_bill'],fill= True)
plt.title("Kde plot of total_bill")
```

```python

# pair plot 

plt.figure(figsize=(5,4))
sbn.pairplot(tips)
```

```python

#heatmap with corr

corre = tips[['total_bill','tip','size']].corr()

sbn.heatmap(corre,annot=True,cmap='coolwarm')

# annot=True → displays the values inside each box.
```
