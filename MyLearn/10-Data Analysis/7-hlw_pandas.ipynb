#### Assignment 3: Data Cleaning

1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python

df = pd.DataFrame(np.random.randint(1,9,size=(5,3)),columns=['A','B','C'])
print(df)
print()

df.iloc[0,1] = np.nan
df.iloc[1,2] = np.nan
df.iloc[4,1] = np.nan
df.iloc[2,2] = np.nan
print("Data Frame with Null value : ")
print(df)
print()
print("Replacing the null value with Mean of columns : ")
df['B'] = df['B'].fillna(df['B'].mean()).astype(int)
df['C'] = df['C'].fillna(df['C'].mean()).astype(int)
print(df)
```

#### Assignment 4: Data Aggregation

1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.


```python

# create pandas dataframe with two columns:

df = pd.DataFrame({'Category':np.random.choice(['A','B','C'],size=(10)),'Value':np.random.randint(1,100,size=(10))})
print(df)
print()

# making group using the function agg 

print("Making group with Category : ")
grouped = df.groupby('Category')['Value'].agg(['mean','sum'])
print(grouped)

# syntax of arg function : (['method-1','method_2']) 

```

2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python

# make df with np -- df.dataframe({'nam':np.random.choice([..]),...,...})

df = pd.DataFrame({'Category':np.random.choice(['Prod1','prod2','prod3'],size=10),'Product':np.random.randint(1,10,size=10),'Sale':np.random.randint(200,600,size=10)})
print(df)
print()
print("Making group by category : ")
grouped = df.groupby('Category')['Sale'].sum()
print(grouped)
```

#### Assignment 5: Merging DataFrames

1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python

df1 = pd.DataFrame({'key':['A','B','C'],'Value':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'],'Value':[4,5,6]})

# merging two dataframe

merged = pd.merge(df1,df2,on='key',how='inner')
print(merged)
```

2. Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.

```python

df1 = pd.DataFrame({'key':['A','B','C'],'Value':[1,2,3]})
df2 = pd.DataFrame({'key':['D','E','F'],'Value':[4,5,6]})

#  merged along with different column

merged = pd.merge(df1,df2,on='key',how='outer')
print(merged)
```

#### Assignment 6: Time Series Analysis

1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

```python

# "Should I create a date every day, every week, every month, etc.?" 'D'

data_range = pd.date_range(start='2021-01-01',end='2021-12-31',freq='D')

df = pd.DataFrame(data_range,columns=['Date'])
df['Data'] = np.random.randint(1,100,size =len(data_range))

df.set_index('Date',inplace=True)
print(df)
print()
print("Monthly mean : ")
monthly_mean =df.resample('ME').mean()
print(monthly_mean)

```

2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

```python

data_rng = pd.date_range(start='2021-01-01',end='2021-12-31',freq='D')

df = pd.DataFrame(data_rng,columns=['Date'])
df['Data'] = np.random.randint(1,100,size=len(data_rng))
df.set_index('Date',inplace=True)
print(df)
print()

print("Rolling window : ")
week_r = df.rolling(window=7).mean()
print(week_r)
```

#### Assignment 7: MultiIndex DataFrame

1. Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python

arr = [['A','A','B','B'],['one','two','one','two']]

multi_ind = pd.MultiIndex.from_arrays(arr,names=('Category','SubCategory'))
df = pd.DataFrame(np.random.randint(1,100,size=(4,3)),index=multi_ind,columns=['Value1','Value2','Value3'])
print(df)

print()
print("Basic Indexing : ")
print(df.loc['A'])
print()
print("slice method : ")
print(df.loc['B','one'])
```

2. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python

arr = [['A','A','A','B','B','B'],['one','two','one','two','one','two']]

mult_i = pd.MultiIndex.from_arrays(arr,names=('Category','SubCategory'))

df = pd.DataFrame(np.random.randint(1,100,size=(6,3)),index=mult_i,columns=['Value1','Value2','Value3'])
print(df)

print()
sum_group = df.groupby(['Category','SubCategory']).sum()
print("Sum of value of Category and SubCategory ")
print(sum_group)
```

#### Assignment 8: Pivot Tables

1. Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.

```python

data_range = pd.date_range(start='2021-01-01',end='2021-01-10',freq='D')

df = pd.DataFrame({'Date':np.random.choice(data_range,size=10),'Category':np.random.choice(['A','B','C'],size=10),'Value':np.random.randint(1,100,size =10)})

print(df)

print()
print("pivot table means :")
pv_table = df.pivot_table(values='Value',index='Date',columns='Category',aggfunc='sum')
print(pv_table)
```

2. Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.

```python


df = pd.DataFrame({'Year':np.random.choice([2000,2010,2020],size=7),'Quarter':np.random.choice(['Q1','Q2','Q3','Q4'],size=7),'Revenue':np.random.randint(1,100,size=7)})

print(df)
print()
pivt = df.pivot_table(values='Revenue',index='Year',columns='Quarter')
print("Pivot table : ")
print(pivt)
```
