
### Pandas in Python 
Pandas is a popular Python library used for data analysis and data manipulation. It helps us organize, clean, and analyze data easily. Pandas provides two main data structures: Series (one-dimensional data) and DataFrame (two-dimensional data with rows and columns, like a table). It can read and write data from files such as CSV and Excel, making it very useful in Data Science, Machine Learning, and Data Analysis.

```python
""" Pandas Series is a one-dimensional labeled array that stores a single column of data.
Each value has an index, which helps access the data easily.
A Series can store numbers, text, or other data types. """


import pandas as pd

data = [10,20,30,40]
series = pd.Series(data)
print(series)
print(type(series))
```

```python

# padans series with dict

data_dict = {'a':2,'b':3,'c':4}
dict_s = pd.Series(data_dict)
print(dict_s)
```

```python
## With own index value
data = [10,20,30,40]
ind = ['a','b','c','d']

series = pd.Series(data,index = ind)
print(series)

```

```python

# data frame from the dictionary 
# dataframe

import pandas as pd

data_dict = {
    'name':["Aayush","Susan","Sabin"],
    'age':[22,32,45],
    'city':["Birtamod","ilam","ktm"]
}

data_frame = pd.DataFrame(data_dict)
print(data_frame)
print(type(data_frame))

```

```python

## accessing data from the table

import pandas as pd

data_dict = {
    'name':["Aayush","Susan","Sabin"],
    'age':[22,32,45],
    'city':["Birtamod","ilam","ktm"]
}
df = pd.DataFrame(data_dict)
print(df)

#  it used present through the level  or Selects data using labels (row/column names).
print()
print(df.loc[0,'name'])
print()
# it used to present from the index
print(df.iloc[1])

# for columhs
df["city"]
```

```python

#  accessing specified element by using at

print(df.at[1,'name'])
print(df.at[2,'age'])

print()

#  accessing specified element by using iat

print(df.iat[0,0])
print(df.iat[2,2])
```

```python

#  data manipulation 

#  adding new colum 

df["Salary"] = [12000,30000,45000]
print(df)

# delete  column axis = 1 for colum and 0 for the row 
# inplace have delete the column permanently 

df.drop('Salary',axis=1,inplace=True)
print()
print(df)

#updating
print()
df["age"] = df["age"]+1
print(df)
```
