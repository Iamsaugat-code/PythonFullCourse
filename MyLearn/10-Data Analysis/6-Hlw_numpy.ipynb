#### Assignment 4: Statistical Operations

#### Assignment 6: Linear Algebra

1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.


```python

arr = np.random.randint(1,10,size=(3,3))

print("Matrix : ")
print(arr)

# calculating det

print()
det = np.linalg.det(arr)
print("Determination of Matrix : ",det)

# inverse

print()
inv = np.linalg.inv(arr)
print("Inverse of matrix : ")
print(inv)

# eigenvalues 

print()
eig = np.linalg.eigvals(arr)
print("Eigenvalues : ")
print(eig)


# note -- (linalg) It is a NumPy module that provides functions for linear algebra and matrix operations.
```

2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

```python

data_f = [('x','i4'),('y','i4')]
data = np.array([(4,6),(9,1),(3,3)],dtype=data_f)
print(data)
print()

#  compute the eucliden distance between the each pair 

distance = np.sqrt((data['x'][:,np.newaxis] - data['x'])**2 + (data['y'][:,np.newaxis] - data['y'])**2)
print("Ecucliden Distance : ")
print(distance)
```

2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python


arr = np.random.randint(1,10,size=(3,3))

masks_arr = ma.masked_array(arr,mask=np.eye(3,dtype=bool))

print(arr)
print()
print(masks_arr)
print()

unmasked_mean = masks_arr.mean()

masks_arr = masks_arr.filled(unmasked_mean)
print(masks_arr)
```
