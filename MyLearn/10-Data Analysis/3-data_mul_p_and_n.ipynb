
### Data manipulation and analysis with Pandas 
means using Python’s Pandas library to clean, organize, modify, and understand data. Pandas helps us read data from files like CSV, select rows and columns, handle missing values, sort data, filter records, and calculate useful information such as averages and totals. It makes working with large datasets easier and faster.

```python

### reading csv file and apply some method

import pandas as pd

df = pd.read_csv("sales.csv")

print("First Five Row :\n")
print(df.head(5))
print()
print("Last Five Row : \n")
print(df.tail(5))
```

```python

#  Data type of table and describe() method 

print("Describe about table : \n")
print(df.describe())
print("\nData type of Table column : \n")
df.dtypes
```

```python

#  managing missing value in table 

# showing table where null value is present 
print(df.isnull())

print()
# showing null in table column 
print(df.isnull().any())

print()
# counting no.of null value in column:
print(df.isnull().sum())
```

```python

#  filling the missing value  with zero 0

null_fill = df.fillna(0)

# filling the missing value with mean() with the new column

df['New_Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())

df['New_Unit_Price'] = df['Unit_Price'].fillna(df['Unit_Price'].mean())


print(df.head(6))
```

```python

df.drop('New_Quantity',axis=1,inplace=True)
df.drop('New_Unit_Price',axis=1,inplace=True)
```

```python

## replacing column name

df =df.rename(columns={'Product':'Items'})
df.head(3)

#  syntax : (columns :{'col_name':'new_col_name'})
```

```python
## changing the data types of column

df['New_Quantity'] = df['Quantity'].fillna(df['Quantity'].mean()).astype(int)
df.head(3)

# note -- changing data type with using(.astype(int,float))
# note -- if there is null then it show error 
```

```python

# Data Aggrigate and grouping
# grouping the items values 

groupby_c= df.groupby('Category')['Quantity'].mean()
print(groupby_c)


# syntax : df.groupby('Column_name')['value_column'].function
```

```python

# grouping by the two object

group_c_i = df.groupby(['Category','Items'])['Quantity'].sum()
print(group_c_i)
print()
group_c_i = df.groupby(['Category','Items'])['Quantity'].mean()
print(group_c_i)


# note  --- (['First_column','Second_column'])['Value_column'].Function()

```

```python

# using muliple arrggiate function 

groupby_arg = df.groupby('Category')['Quantity'].agg(['sum','mean','count'])
print(groupby_arg)

# note -- some of the .agg(['sum',;......,'..'])
```

```python

# apply the function in data frame column

df['New_Quantity']=df['Quantity'].fillna(df['Quantity'].mean()).astype(int).apply(lambda x :x*2)
df.head(3)


# note -- changing one dataype column to another data type (.astype(data type))
# note -- if there is null in column then function cannot be apply

```

```python

## merging two dataframe 

df1 = pd.DataFrame({'key':['A','B','C'],'value_1':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'],'value_2':[3,4,5]})

print(pd.merge(df1,df2,on='key',how='left'))

# how == inner(come),outer(all),left(a-b),right(b-a)
```


### Reading data from the various source using pandas 

 StringIO is used to treat a string like a file.
 This is useful because pd.read_json() can read JSON from a file-like object.
 Your JSON contains multiple lines, so triple quotes allow you to write it across several lines.


```python

import pandas as pd

from io import StringIO


data = '''{
    "Employe_Name":"Sabin Rai",
    "Email":"raisabin@gmail.com",
    "Profile":[{"title_1":"Team Leader","title_2":"Developer"}]}'''

rd = pd.read_json(StringIO(data))
rd

```

```python
# again changing into json

print(rd.to_json())
print()
# with index default and it became the key
print(rd.to_json(orient='index'))
print()
# with orient = records and it show the data as object
print(rd.to_json(orient='records'))
```
