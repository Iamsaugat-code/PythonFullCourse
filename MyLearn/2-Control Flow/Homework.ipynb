### Assignment 3: if-elif-else Statement

Write a program that asks the user to input a number and prints whether the number is positive, negative, or zero.
 and check odd or even

```python
num = int(input(" Enter the Number : "))
if num >0:
    print(f" {num} is a positive number")
    if num%2==0:
        print(f"{num} is a even number")
    else:
        print(f"{num} is a odd number")
elif num <0 :
    print(f"{num} is a negative number")
    if num%2==0:
        print(f"{num} is negative even number")
    else:
        print(f"{num} is a negative odd number")
else:
    print(f"{num} is  zero ")
```

### Assignment 8: break Statement

Write a program that asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.


```python

#  making sum of the all input number until the user print a 0 ;

sum =0;
while 1:
    num = int(input(" Enter the Number : (0) means exit()"))
    if num == 0:
        break
    sum = sum +num

print(f"Sum of input number is {sum}") 
```

## Assignment 9: continue Statement

Write a program that prints all the numbers from 1 to 10 except 5 using a for loop and continue statement.


```python
#  making a code with pass and continue statement

for i in range(1,5):
    if i ==3:
        continue
    if i ==2:
        print("pass i :",i)
        pass
    print(i)
```

### Assignment 12: Factorial Calculation

Write a program that calculates the factorial of a number input by the user using a while loop.


```python

#  making factorial question practising 

fact =1
n = int(input(" Enter the Number :"))
count =1
while count <=n:
    fact *= count
    count += 1
print(f" factorial of {n} is {fact}")
```

## Assignment 14: Prime Number Check

Write a program that checks if a number input by the user is a prime number using a for loop.


```python
# making even number
n = int(input(" Enter the number for to check the prime number or not : "))
is_prime = True
if n<=1:
    print(" invalid input : ")
else:
    for i in range(2,int(n ** 0.5)+1):
        if n %i ==0:
            is_prime = False
            break
if is_prime:
    print(f"{n} is a prime number :")
else:
    print(f"{n} is not a prime number : ")
```

# Assignment 15: Fibonacci Sequence

Write a program that prints the first n Fibonacci numbers, where n is input by the user.

```python

#  makeing 0 1 1 2 3  as output

n = int(input(" Enter the Number : "))
a,b = 0,1
count = 0
while count <=n:
    print(a)
    temp = a
    a =b
    b = temp +a
    count += 1
```
