```python

## making calculator using if elif and else condition

a = int(input(" Enter the first number : "))
b = int(input(" Enter the second number : "))
c = (input(" Enter the Operator (*,%,/,+,-)"))

if c == '+':
    print("Addition : ",a+b)
elif c =='-':
    print("Subtraction : ",a-b)
elif c =='*':
    print("Multiplication : ",a*b)
elif c =='/':
    if a == 0:
        print("  Cannot Divide ( 0 / B )")
    else :
        print(" Division : ",a/b)
elif c=='%':
    if a == 0:
        print(" Invalid method")
    else:
        print(" Moduls : ",a%b)
else:
    print(" invalid input : ")
```
