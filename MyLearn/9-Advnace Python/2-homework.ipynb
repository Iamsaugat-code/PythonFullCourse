#### Assignment 1: Custom Iterator

Create a custom iterator class named `Countdown` that takes a number and counts down to zero. Implement the `__iter__` and `__next__` methods. Test the iterator by using it in a for loop.

```python

class Countdown:
    
    def __init__(self,number):
        self.number = number
    
    # “This object can be used in a for loop.”
    
    def __iter__(self):
        return self
    
    # So, the Countdown object itself works as an iterator (when we return self).
    
    
    # The __next__() method tells Python how to get the next value.
    
    def __next__(self):
        if self.number <=0:
            raise StopIteration
        # Python keeps calling __next__() until StopIteration occurs.
        
        else :
            self.number -= 1
            return self.number

for i in Countdown(6):
    print(i,end=" ")
    
    
```

#### Assignment 2: Custom Iterable Class

Create a class named `MyRange` that mimics the behavior of the built-in `range` function. Implement the `__iter__` and `__next__` methods. Test the class by using it in a for loop.

```python

class Myrange:
    
    def __init__(self,start,end):
        self.start = start
        self.end = end
    
    def __iter__(self):
        return self
    
    def __next__(self):
        
        if self.start >= self.end:
            raise StopIteration
        else:
            self.start += 1
            return self.start


for i in Myrange(0,6):
        print(i,end=" ")

        
```

#### Assignment 3: Generator Function

Write a generator function named `fibonacci` that yields the Fibonacci sequence. Test the generator by iterating over it and printing the first 10 Fibonacci numbers.

```python

def fibo(n):
    a,b=0,1
    for _ in range(n):
        yield a
        a,b=b,a+b

for i in fibo(10):
    print(i,end=" ")
```

```python

def even_num(a,b):
    for i in range(a,b):
        if i%2 == 0:
            yield i*i

def square(a):
    for i in a:
        yield i*i


num = even_num(1,21)
sq = square(num)

for i in sq:
    print(i,end=" ")


   
    
```

#### Assignment 6: Simple Decorator

Write a decorator named `time_it` that measures the execution time of a function. Apply this decorator to a function that calculates the factorial of a number.

```python

import time

def time_it(func):
    ans = 0
    def wrap(*arg,**kwarg):
        time_start = time.time()
        result = func(*arg,**kwarg)
        time_end = time.time()
        print(f"Time execution : {time_start-time_end} second ")
        return result
    return wrap

@time_it
def fact(n):
    if n ==0:
        return 1
    else:
        return n*fact(n-1)

print(fact(3))
```

#### Assignment 7: Decorator with Arguments

Write a decorator named `repeat` that takes an argument `n` and repeats the execution of the decorated function `n` times. Apply this decorator to a function that prints a message.

```python
"""
*arg and **kwarg make the decorator work with any function, 
no matter how many arguments that function has."""

def repeat(n):
    def decorator(func):
        def wrap(*arg,**kwarg):
            for _ in range(n):
                func(*arg,**kwarg)
        return wrap
    return decorator

@repeat(2)
def say_hello(msg):
    print(msg)

say_hello("i am sabin rai ")
        
```

#### Assignment 8: Nested Decorators

Write two decorators: `uppercase` that converts the result of a function to uppercase, and `exclaim` that adds an exclamation mark to the result of a function. Apply both decorators to a function that returns a greeting message.

```python

def uppercase(func):
    def wrap(*arg,**kwarg):
        result = func(*arg,**kwarg)
        return result.upper()
    return wrap

def exclaim(func):
    def wrap(*arg,**kwarg):
        result = func(*arg,**kwarg)
        return result + "!"
    return wrap

@uppercase
@exclaim
def say_hello(ms):
    return f"hello : {ms}"

print(say_hello("saugat rai"))
```

#### Assignment 10: Iterator Protocol with Decorators

Create a custom iterator class named `ReverseString` that iterates over a string in reverse. Write a decorator named `uppercase` that converts the string to uppercase before reversing it. Apply the decorator to the `ReverseString` class.

```python

def decorator(cls):
    class Wrap(cls):
        def __init__(self,*arg,**kwarg):
            super().__init__(*arg,**kwarg)
            self.name = self.name.upper()
    
    return Wrap

# ReverseString = uppercase(ReverseString) we can write in this way also

@decorator
class ReverseString:
    def __init__(self,name):
        self.name = name
        self.ind = len(name)
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.ind == 0:
            raise StopIteration
        else:
            self.ind -= 1
        return self.name[self.ind]

for i in ReverseString("Hello"):
    print(i,end=" ")
        
```

Write a generator function named `safe_divide` that takes a list of numbers and yields the division of each number by a given divisor. Implement exception handling within the generator to handle division by zero.

```python
def safe_divide(lst,div):
  try :
      if div ==0:
          raise ZeroDivisionError
      else:
          for i in lst:
              yield i /div
  except ZeroDivisionError:
      print("Divide by Zero :")
      
      
for i in safe_divide([2,4,6,8,10],2):
    print(i,end=" ")
```

### Assignment 13: Context Manager Decorator

Write a decorator named `open_file` that manages the opening and closing of a file. Apply this decorator to a function that writes some text to a file.

```python

def open_file(file_name,mode):
    def decorator(func):
        def wrap(*arg,**kwarg):
            with open(file_name,mode) as file:
                return func(file,*arg,**kwarg)
        return wrap
    return decorator

@open_file("exmple.txt","w")
def file_to_write(file,text):
    file.write(text)

file_to_write("Hi i am hulk rai")
```

#### Assignment 14: Infinite Iterator

Create an infinite iterator class named `InfiniteCounter` that starts from a given number and increments by 1 indefinitely. Test the iterator by printing the first 10 values generated by it.

```python

class InfiniteCounter:
    def __init__(self,start =0 ):
        self.start = start
    
    def __iter__(self):
        return self
    
    def __next__(self):
        self.start += 1
        return self.start


c = InfiniteCounter(10)
for _ in range(10):
    print(next(c),end=" ")
        
```

### Assignment 15: Generator Pipeline

Write three generator functions: `integers` that yields integers from 1 to 10, `doubles` that yields each integer doubled, and `negatives` that yields the negative of each doubled value. Chain these generators to create a pipeline that produces the negative doubled values of integers from 1 to 10.

```python

def integer():
    for i in range(1,11):
        yield i

def doubles(nums):
    for i in nums:
        yield i+i

def negative(nums):
    for i in nums:
        yield -i


int_v = integer()
doub = doubles(int_v)
neg = negative(doub)

for va in neg:
    print(va,end=" ")
```
