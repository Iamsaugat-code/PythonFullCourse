
### Generator 

Generator is the simple way to create iterator. They use the yeild keyword for to produce a series of values lazily which means they generate values on the fly and not to store in memory

```python

#simle example of yeild 

def sqaure(n):
    for i in range(n):
        yield i*i

print(sqaure(3))
a = sqaure(3)
```

```python
## with the next function

next(a)
```

```python
## with for loops
for i in sqaure(3):
    print(i,end=" ")
```

```python

## generating yeild 

def generate_y():
    yield 1
    yield 2
    yield 3

a = generate_y()
```

```python
next(a)
```

#### Practical exmample of generator 


```python

def read_file(filepath):
    with open(filepath,"r") as file:
        for line in file:
            yield line

```

```python
file_path = "exmple.txt"

for i in read_file(file_path):
    print(i.strip())
```


### Decorator 
A decorator is a design pattern or programming feature that allows you to add new functionality to a function, method, or class without modifying its original code.
In Python, a decorator is a function that takes another function as input, extends or changes its behavior, and returns the modified function.

```python

## Copy method 

def welcome():
    return "Welcome to the decorator class : "

wel = welcome
print(wel())
del welcome
print(wel())

# note well copying we should not made "()" only we should made function name
```

```python

## closures with function :

def welcomes(msg):
    
    def inside_welcome():
        print("This is the first class:")
        print(msg)
        print("This is the last class")
    
    return inside_welcome()

welcomes("and")

# note --- while creating copy we should make only function obj without this "()"
```

```python

## Closure with function
## print function 

def welcomes(func):
    
    def inside_welcome():
        print("This is the first class:")
        func("and")
        print("This is the last class")
    
    return inside_welcome()

welcomes(print)

#  note - we can send function as parameter with func
```

```python
## closures with another example:

def welcomes(func,lst):
    
    def inside_welcome():
        print("This is the first class:")
        print(func(lst))
        print("This is the last class")
    
    return inside_welcome()

welcomes(len,[1,2,3,4,5])

# note - we can send the function as parameter and data types also 

```

```python

## decorator with function

def welcomes(func):
    
    def inside_welcome():
        print("This is the first class:")
        func()
        print("This is the last class")
    
    return inside_welcome()

@welcomes
def say_hello():
    print("this is the middle one : ")
    

# note -- while creating decorator we should use @function_name without this"()"
```

```python

# one more example of decorator 

def my_func(func):
    def wrapper():
        print("Hi am diwash : ")
        func()
        print("Hi am pujan : ")
    
    return wrapper

@my_func
def say_hello():
    print("Hi iam saugat ")

say_hello()
```

```python

# decorator with parameter 

def repeat(n):
    def decorator(func):
        def wrape(*arg,**kwarg):
            for _ in range(n):
                func(*arg,**kwarg)
        
        return wrape
    return decorator

# note -- if there is parameter than we should make this () otherwise no need to make this 

@repeat(3)
def say_hello():
    print("Hi i am iron man : ")

say_hello()

# Why use *arg and **kwarg if say_hello() has no arguments?
```
