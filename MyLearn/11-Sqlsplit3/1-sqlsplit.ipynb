
### SQL and SQLSplit
SQL (Structured Query Language) is used to create, manage, and retrieve data from databases using commands like SELECT, INSERT, UPDATE, and DELETE. SQLSplit is used to split a large SQL script or query into separate SQL statements. The main difference is that SQL is a database query language, while SQLSplit is used to separate SQL statements<br>
The cursor performs the SQL operation, while commit() saves the changes permanently

```python

# Inputing the library  

import sqlite3
```

```python

# making connection the database

connection = sqlite3.connect("Example.db")
connection

```

```python

# creating a cursor

cursor = connection.cursor()
 
```

```python

# creating table 

cursor.execute('''
               
               create table if not exists Employes(
                  ID integer primary key,
                  Name text not null,
                  Age text,
                  Department text 
               )
               ''')

connection.commit()
```

```python

#  inserting a data into the table 

cursor.execute('''
               
               insert into Employes(Name,Age,Department) values('Sabin Rai',32,'Science')
               
               ''')

cursor.execute('''
               
               insert into Employes (Name,Age,Department) values('Suman Rai',44,'Artist')
               
               ''')

cursor.execute('''
               
               insert into Employes (Name,Age,Department) values('Sabin Rai',34,'Management')
               
               ''')

connection.commit()
```

```python

#  updating the data in the table 

cursor.execute('''
               
               update Employes 
               set name = 'susan Rai'
               where age = 34
               ''')

connection.commit()
```

```python

#  deleting the data from the table

cursor.execute('''
               
               delete from employes
               where name = 'susan Rai'
               
               ''')

connection.commit()
```

```python

#  displaying the data from the database

cursor.execute('select * from Employes')
rows = cursor.fetchall()

for row in rows:
    print(row)
    
```

```python

#  closing the connection of the database

connection.close()
```

```python

# droping the table 

cursor.execute('drop table market')
connection.commit()
```
