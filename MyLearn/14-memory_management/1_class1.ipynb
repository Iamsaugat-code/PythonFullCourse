### Memory Management :
Memory management in Python is the process of allocating and freeing memory while a program is running. Python manages memory automatically, so the programmer usually does not need to free memory manually
1) Reference Counting – Python keeps track of how many references point to an object. When no reference remains, the memory can be released.
2) Garbage Collection – Python automatically finds and removes unused objects, especially objects involved in circular references.


#### Reference Count 

Reference count means the number of references (variables or names) pointing to the same object in memory. When the reference count becomes 0, Python can remove that object from memory.

```python

## reference count

import sys 

a = [1,2]

print(sys.getrefcount(a))


# a is one reference to the list.
# When you pass a to sys.getrefcount(a),
# Python temporarily creates another reference as the function argument.
```

```python
b = a
print(sys.getrefcount(b))

# b reference will be added
```

```python

del b
print(sys.getrefcount(a))
```


#### Garbage Collection
It is a Python mechanism that automatically finds and removes unused objects from memory.

Python's garbage collection is especially useful for circular references.

```python

import gc

# making garbage collection enable

gc.enable()
```

```python

# making disable garbage 

gc.disable()
```

```python
#  making garbage value collection 

gc.collect()

# it means the garbage collector collected 1048 unreachable objects at that moment.
```

```python

#  garbage collection start 

print(gc.get_stats())
```

```python

# reaching unreachable garbage

print(gc.garbage)
```


#### Best Practices for Memory Management in Python
1. Use Local Variables
Use variables only where they are needed so they can be removed from memory after the function finishes.
2. Avoid Circular References
Avoid objects unnecessarily referring to each other because they can make memory cleanup more difficult.
3. Use Generators
Generators process data one item at a time, which uses less memory than storing all data at once.
4. Explicitly Delete Objects
Use del to remove a reference to an object when it is no longer needed. This can help free memory.
5. Profile Memory Usage
Check how much memory your program uses to find memory-heavy parts and optimize them.

```python

# example of cicrular reference 

import gc 

class Myobj:
    def __init__(self,name):
        self.name = name
        print(f"{self.name} is created : ")
    
    def __del__(self):
        print(f"{self.name} is delete : ")
        

# reference circular :

obj1 = Myobj("obj1")
obj2 = Myobj("Obj2")

obj1.ref= obj2
obj2.ref= obj1

del obj1
del obj2

gc.collect()
```

```python


# using generator of efficient memory used 

def generator(n):
    for i in range(n):
        yield i

for num in generator(10000):
    print(num,end=" ")
    if num >10:
        break
    
```

```python


# profile memory used 

import tracemalloc

def create_lst():
    return [i for i in range(10000)]

def main():
    
    tracemalloc.start()
    create_lst()
    snapshot = tracemalloc.take_snapshot()
    start_top = snapshot.statistics('lineno')
    
    
    print("[Top ten ]")
    
    for i in start_top[::]:
        print(i)

        
```
