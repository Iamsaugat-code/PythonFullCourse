```python

#  set mathematical operation

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

union_set = set1.union(set2) # union set help to print all element in to set
print(" Union_Set : ",union_set)

intersection_set = set1.intersection(set2)   # element should be in both
print(" Interection_set : ",intersection_set)

set1.intersection_update(set2)  # it will update the set with the value of iin.. set1 and set 2
print(" new set 1 : ",set1)
```

```python
#  checking for difference

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

print(set1.difference(set2))
print(set1)

#  checking for issuper set and subset

set1= {1,2,3,4,5}
set2={3,4,5}

print( set2.issubset(set2))   # set 2 is child of set1
print(set1.issuperset(set2))  # set1 is more bigger than set 2

```

# Dictionaries lesson
<p> dictionaries is the collection of different data combination  with key value 

```python
#  first creating a dictionaire
#  their type 
dictionary = {}
print(type(dictionary))
dictionary_1 = dict()    # using a key word(dict) we are making a dictionary data
print(type(dictionary_1))

dictionary_data = {"name":" Ram karki", "age":34,"grade":'A'}
print(dictionary_data)
print(dictionary_data["name"])          # method one
print(dictionary_data.get("grade"))     # method two

#  note if we use same kind of key more than same last key data will print

print(dictionary_data.get("section"))   # this the right way to know the undefine key 
```

```python
#  modifying ,deleting and updating the key value

dictionary_data = {"name":" Ram karki", "age":34,"grade":'A'}

#  this updating a data
dictionary_data["age"] =45
print(dictionary_data)

#  this is adding a data
dictionary_data["section"] ='d'
print(dictionary_data)

#  this is a deleting a data
del dictionary_data["grade"]
print(dictionary_data)

print("\n")
#  making dictionaries key print
dict_key = dictionary_data.keys()             # .keys() help to print the key of dictonirer
print(" dictionarie value : ",dict_key)

#  making dictionaries value print
dict_val = dictionary_data.values()           # .values() same as above
print(f" dicationarie : {dict_val}")

# making paris

pair_dict = dictionary_data.items()            #.items() help to print all data
print(" Pairs of dictionaries : ",pair_dict)
```

```python
#  shallow copy 

#  checking error
dictionary_data = {"name":" Ram karki", "age":34,"grade":'A'}


dict_copy = dictionary_data
#  updating dictionary_data
dictionary_data["name"] ="sita rai"
print(dictionary_data)
print(dict_copy)

#  if one updation another also automatically update. so this is the error

print(f"\n after using shallow copy method : \n")

dict_copy1 = dictionary_data.copy()            # .copy() is the method for maintaion error
dictionary_data["name"] ="bishal rai"
print(dictionary_data)
print(dict_copy1)


```

```python
#  ierating a loops in dicionaries
dictionary_data = {"name":" Ram karki", "age":34,"grade":'A'}
dict_copy1 = dictionary_data.copy() 
for keys,vals in dict_copy1.items():
    print(f"{keys} : {vals} ")
   
   
#  for nested loops in dictionaris

std = {
    "std_1":{"name" :" sital rai ","age":25},
    "std_2":{"name":"Kumar rai","age":22}
}
print("printing a nested dict _data : ",std["std_1"]["name"])
print("data _ 2 Age : ",std["std_2"]["age"])

#  note for to print we should write the key name for to print the data

#  now printing a data by loop
print()
print("For nested iterating loop : \n")
for std_i , std_info in std.items():
    print(f"{std_i} : {std_info}")
    for keys,val in std_info.items():
        print(f"{keys} : {val}")
        
    

```

```python
#  dictionary comprehension 

dict_1 = {i:i*i for i in range(1,6)}
print(" simple : ",dict_1)

dict_2 = {i : i*i for i in range(1,6) if i%2==0}
print(" Conditional Method : ",dict_2)
```

```python
#  one practise question 
#  finding a frequency of element for list and return in dictionary

lst = [1,2,3,4,5,5,2,1]
freq= {}
for i in lst:
    if i in freq:
        freq[i] = freq[i]+1
    else:
        freq[i] =1

print(freq)
    
```

<h4> calculating the grade of ram 

```python

ram = [67,77,87,99,25]
ram.append(82)
ram.reverse()
print()
print(" Grade  of ram : ",ram)

avg_grade = sum(ram)/len(ram)
print(" Average Grade of Ram : ",avg_grade)

max_grade = max(ram)
min_grade = min(ram)

print(f" Highest Grade of Ram is {max_grade}")
print(f" Lowest Grade of Ram is {min_grade}")

print()
print(" Displaying all grade of Ram : ")
for ind,i in enumerate(ram):
    print(f" Sub {ind+1}  =  {i}")
```
