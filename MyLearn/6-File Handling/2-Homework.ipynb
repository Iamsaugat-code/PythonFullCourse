### Assignment 1: Reading a File

Write a function that reads the contents of a file named `sample.txt` and prints each line.

```python

def read_from_file(filepath):
    with open(filepath,'r') as file:
        for line in file:
            print(line.strip())
        
read_from_file("Example.txt")
```

### Assignment 2: Writing to a File

Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

```python

def list_to_file(path,lst):
    with open(path,'w') as file:
        file.writelines(lst)
        
lst = ['apple \n','Mango \n','Orange \n','Banana ']

list_to_file("Output.txt",lst)
```

### Assignment 3: Copying a File

Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

```python

def copying_data(source,destination):
    with open(source,'r') as file:
        contents = file.read()
    
    with open(destination,'w') as file:
        file.write(contents)

copying_data("Output.txt","Example.txt")
```

### Assignment 5: Counting Words in a File

Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
def couting_words(path):
    with open(path,'r') as file:
        lines= file.readlines()
        count_words = sum(len(line.split()) for line in lines)
    return count_words

words = couting_words("Example.txt")
print(f" No. of words present in file is : {words}")
```

### Assignment 6: Finding and Replacing Text

Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
def find_and_replace(path,old,new):
    with open(path,'r') as file:
        words = file.read()
        new_words = words.replace(old,new)
    with open(path,'w') as file:
        file.write(new_words)

find_and_replace("Output.txt","apple","Messi")
```

### Assignment 7: Reading a File in Reverse

Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.


```python

def reversing(path):
    with open(path,'r')as file:
        lines = file.readlines()
    for line in reversed(lines):
        print(line.strip())

reversing("Output.txt")
        
```

### Assignment 8: Counting Lines, Words, and Characters

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

```python
def couting_data(path):
    with open(path,'r') as file:
        lines = file.readlines()
        count_line = len(lines)
        count_words  = sum(len(line.split()) for line in lines)
        count_character = sum(len(line) for line in lines)
    
    return count_line,count_words,count_character

line,words,character = couting_data("destination.txt")
print(f"A file Has {line} line, {words} words, {character} character ")
```

### Assignment 9: Merging Multiple Files

Write a function that merges the contents of multiple files into a single file named `merged.txt`.

```python

def merged_file (file_lst,merged_file):
    with open(merged_file,'w') as outfile:
        for f_name in file_lst:
            with open(f_name,'r') as infile:
                outfile.write(infile.read() +'\n')

merged_file(["Example.txt","Output.txt"],"destination.txt")
```

### Assignment 10: Splitting a Large File

Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

```python
def split_large(filename,line_per_file):
    with open(filename,'r') as file:
        lines = file.readlines()
    for i in range(0,len(lines),line_per_file):
        with open(f'{filename}_part{i//line_per_file+1}','w') as file_part:
            file_part.writelines(lines[1:1 + line_per_file])

split_large("large.txt",100)
            
```

### Assignment 11: Creating a Log File

Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

```python

import datetime

def log_file(message,filename="activity.log"):
    timestamps = datetime.datetime.now().isoformat()
    with open(filename,'a') as file:
        file.write(f'\n[{timestamps}],{message}')

log_file("i love you too")
```

### Assignment 13: CSV File Operations

Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

```python
import csv

def csvfile(filename):
    with open(filename,'r') as file:
        contents = csv.DictReader(file)
        return list(contents)

print(csvfile("data.csv"))
```

### Assignment 14: JSON File Operations

Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary.

```python

import json

def reading_json(filename):
    with open(filename,'r') as file:
        data = json.load(file)
        return data
print(reading_json("Data.json"))
```


### Assignment 15: File Permission Handling

Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

```python

def checking_p(filename):
    try:
        with open(filename,'r') as file:
            print(file.read())
    except PermissionError as e:
        print(f" Premission error : {e}")

checking_p("Output.txt")
```
