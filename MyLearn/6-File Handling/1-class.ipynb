### file Handling - operation (read and write)

file handling cruical part of the any programming langauge. It provide built-in-function and provide the option of read and write operation in binary or text form

```python
''' Reading From the File '''

with open('Example.txt','r') as file:
    content = file.read()
    print(content)
```

```python
''' reading data from the file line by line '''

with open('Example.txt','r') as file:
    for line in file:
        print(line.strip())

#  note - line.strip() help to remove '\n' while printing line by line 
```

```python
''' Writing in file '''

with open('Example.txt','w') as file:
    file.write("\nHello Hello ! \n")
    file.write("This is the message from the Apple store : \n")
    print(" Successfully Written in File : ")
```

```python
''' Using Append mode for to write Data in file '''


with open('Example.txt','a') as file:
    file.write("This is message from append operation : \n")
    print(" Successfully Written : ")
    
#  note - append means written with past Data
```

```python
''' Writing multiple line in file '''

data = ['First line \n','Second line \n','Thrid line \n']

with open("Example.txt",'a') as file:
    file.writelines(data)
    print(" Successfully ")
```

```python

''' Writing in binary form in file '''

bin_data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(bin_data)
```

```python
''' Reading Binary data from the file : '''

with open('example.bin','rb') as file:
    content = file.read()
    print(content)

### note - file.read() means it read all line in file 
```

```python
''' Copy from the source file and write in destination file '''

#  copying file data to another file 

with open('Example.txt','r') as source:
    content_1 = source.read()

#  Writing Destination file 

with open('destination.txt','w') as file:
    file.write(content_1)
```

```python
''' Counting line words and character in file '''

def file_data(filepath):
    with open(filepath,'r') as file:
        lines = file.readlines()  # note - it means read data line by line 
        count_line = len(lines)
        count_words = sum(len(line.split()) for line in lines)
        count_character = sum(len(line) for line in lines)
        return count_line,count_words,count_character

filepath = 'Example.txt'

line,words,character = file_data(filepath)
print()
print(f"Lines : {line} , Words : {words} , characters : {character}")
        
```

```python
## Reading and writting operation in file 

## note - w+ allow the user to write and read file

with open('Example.txt','w+') as file:
    file.write("Hello Hello ! \n")
    file.write("Hi i am from new line : \n")
    
    #  note - file.seek(0) help to move cursor to the beggining of the file
    
    file.seek(0)
    
    # without it we cannot read the lines in file 
    
    contents = file.read()
    print(contents)
```

#### Working in different file path 

```python
## making new direction 

import os 
new_dir = "package"
os.mkdir(new_dir)

print(f" New Direction is create : ")
```

```python
## showing list of file in this folder
import os 

print(os.listdir('.'))

# note ('.') - means list of folder or file in this folder
```

```python
#  joining Folder inside file 

import os 

folder_name = "Folder"
file_name = 'sure.txt'
print(os.path.join(os.getcwd(),folder_name,file_name))
```

```python

''' Checking path is present or not '''

import os 

file_name = "Example.txt"
if os.path.exists(file_name):
    print(f"{file_name} is present : ")
else:
    print(f"{file_name} is not present : ")
```

```python
''' Checking is this path is file,folder nor '''

import os 

file_name = "Example.txt"
if os.path.isfile(file_name):
    print(f"{file_name} is the file : ")
elif os.path.isdir(file_name):
    print(f"{file_name} is the Directory : ")
else:
    print(f"{file_name} is not file or folder : ")
```

```python
## making absulote path 

relative_path = "Example.txt"
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
